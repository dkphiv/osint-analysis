```python
# OSINT Social Media Analysis

Ziel dieser Analyse ist die Untersuchung öffentlicher Textdaten (z. B. Social Media), um zentrale Themen, Muster und Trends zu identifizieren.

Der Fokus liegt auf:
- Datenbereinigung
- Explorative Analyse
- Keyword-Extraktion (TF-IDF)
- Visualisierung von häufigen Begriffen
```

```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Pfad anpassen!
df = pd.read_csv("data/comments.csv")

# Nur relevante Spalte behalten
df = df[['body']].dropna()

print("Datensätze:", len(df))
df.head()
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^a-zA-ZäöüÄÖÜß\s]", "", text)
    return text

df["clean"] = df["body"].apply(clean_text)

df.head()
```

```python
words = " ".join(df["clean"]).split()
counter = Counter(words)

top_words = counter.most_common(15)

labels, values = zip(*top_words)

plt.figure(figsize=(10,5))
plt.bar(labels, values)
plt.xticks(rotation=45)
plt.title("Top 15 häufigste Wörter")
plt.show()
```

```python
vectorizer = TfidfVectorizer(max_features=15, stop_words='english')
X = vectorizer.fit_transform(df["clean"])

keywords = vectorizer.get_feature_names_out()

print("Wichtige Keywords:")
print(keywords)
```

```python
## Interpretation der Ergebnisse

Die Analyse zeigt, welche Begriffe in den untersuchten Textdaten besonders häufig oder relevant sind.

Mögliche Erkenntnisse:
- Identifikation dominanter Diskussionsthemen
- Hinweise auf Trends oder Stimmungen
- Grundlage für weiterführende Analysen (z. B. Klassifikation oder Netzwerkanalyse)

Diese Art der Auswertung kann im OSINT-Kontext genutzt werden, um frühzeitig relevante Entwicklungen in öffentlichen Datenquellen zu erkennen.
```
